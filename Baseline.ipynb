This is just the syntax for a pytorch model. It establishes each layer we mentioned in the powerpoint

```python
class Baseline(nn.Module):
    def __init__(self):
        super().__init__()

        self.Conv2_1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, stride=1, padding=1)
        self.activation = nn.ReLU()
        self.MaxPool1 = nn.MaxPool2d(kernel_size=2, stride=2)
        self.Conv2_2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1, padding=1)
        self.activation2 = nn.ReLU()
        self.MaxPool2 = nn.MaxPool2d(kernel_size=2, stride=2)
        self.Flatten = nn.Flatten()
        self.Dense = nn.Linear(in_features=4096, out_features=256)
        self.activation3 = nn.ReLU()
        self.OutPut = nn.Linear(in_features=256, out_features=100)

    
    def forward(self, x):
        x = self.Conv2_1(x)
        x = self.activation (x)
        x = self.MaxPool1(x)
        x = self.Conv2_2(x)
        x = self.activation2(x)
        x = self.MaxPool2(x)
        x = self.Flatten(x)
        x = self.Dense(x)
        x = self.activation3(x)
        x = self.OutPut(x)   
        return x
```
